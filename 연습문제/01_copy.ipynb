## 1-1. Daum 뉴스기사 제목 스크래핑하기 
* 아래의 url에서 뉴스기사의 링크와 제목을 출력하기
    - 다음 경제 뉴스 URL
    url = 'https://news.daum.net/economy'

```python
import requests
from bs4 import BeautifulSoup

# 다음 경제 뉴스 URL
url = 'https://news.daum.net/economy'

print(url)

# 요청 헤더
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/104.0.0.0 Safari/537.36'
}

res = requests.get(url, headers=req_header)
print(type(res))
print(res.status_code)

# 정상 응답 여부 확인
if res.ok:
    res.encoding = 'utf-8' 
    html = res.text
    soup = BeautifulSoup(html, 'html.parser')
    
    # 기사 제목과 링크 추출
    li_tag_list = soup.select('ul.list_newsheadline2 li')

    print(type(li_tag_list), len(li_tag_list))

    for li_tag in li_tag_list:        
        a_tag = li_tag.find('a')
        print(a_tag['href'])
        
        #strong_tag = li_tag.select('div.cont_thumb strong.tit_txt')[0]
        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        title = strong_tag.text.strip()
        print(title)

else:
    print(f'에러코드 = {res.status_code}')
```

## 1-2. Daum 뉴스기사 제목 스크래핑하기 코드를 섹션별로 처리하는 함수로 구현하기
* 여러개의 section 중 하나를 선택해서 url에서 뉴스기사의 링크와 제목을 출력하는 코드를 함수로 작성하기
    1) 경제 뉴스 URL
    url = 'https://news.daum.net/economy'
    2) 사회 뉴스 URL
    url = 'https://news.daum.net/society'
    > 사회 section 선택



```python
import requests
from bs4 import BeautifulSoup

def print_news(section_name):    
    # selection_dict['사회] => society
    section = section_dict[section_name]
    # 요청 Parameter
    req_param = {
        'section': section
    }
    url = 'https://news.daum.net/{section}'.format(**req_param)
    
    print(f'======> {url} {section_name} 뉴스 <======')
    
    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/104.0.0.0 Safari/537.36'
    }

    res = requests.get(url, headers=req_header)   

    if res.ok:
        res.encoding = 'utf-8' 
        html = res.text
        soup = BeautifulSoup(html, 'html.parser')
        
        # 기사 제목과 링크 추출
        li_tag_list = soup.select('ul.list_newsheadline2 li')

        #print(type(li_tag_list), len(li_tag_list))

        for li_tag in li_tag_list:        
            a_tag = li_tag.find('a')
            print(a_tag['href'])
            
            strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
            title = strong_tag.text.strip()
            print(title)


    else:
        print(f'에러코드 = {res.status_code}')
```

## 2-1. Nate 뉴스기사 제목 스크래핑하기 (선택)
* 최신뉴스, 정치 , 경제, 사회, 세계, IT/과학 
6개의 섹션의 뉴스를 출력하는 함수를 생성하여 스크래핑 하기
- Nate url: https://news.nate.com/recent?mid=n0100
- Image, 기사제목, 기사링크


```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from IPython.display import Image, display

url = 'https://news.nate.com/recent?mid=n0100'
print(url)

req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/104.0.0.0 Safari/537.36'
}

res = requests.get(url, headers=req_header)
print(type(res))
print(res.status_code)

if res.ok:
    res.encoding = 'euc-kr'
    html = res.text
    soup = BeautifulSoup(html, 'html.parser')
    
    tags = soup.select("div.postListType.noListTitle div.mlt01")
    print(len(tags))

    for idx,div_tag in enumerate(tags,1): 
        print(f'============>> {idx}')       
        a_tag = div_tag.find('a')
        a_join_url = urljoin(url,a_tag['href'])
        print(f'뉴스기사 링크 = {a_join_url}')

        img_tag = div_tag.select_one('span.ib img')
        if img_tag:
            photo_url = urljoin(url,img_tag['src'])
            print(img_tag['src'])
            print(photo_url)
            display(Image(url=photo_url))

        h2_tag = div_tag.select_one('span.tb h2.tit')
        title = h2_tag.text
        print(title)

else:
    print(f'에러코드 = {res.status_code}')
```

## 2-2. 하나의 네이버 웹툰과 1개의 회차에 대한 Image 다운로드 하기 (필수)

```python
import os
import requests
from bs4 import BeautifulSoup

def download_one_episode(title, no, url):
    # 디렉토리 경로 생성 (img/일렉시드/341)
    save_dir = os.path.join('img', title, str(no))
    os.makedirs(save_dir, exist_ok=True)
    
    # Headers 설정
    req_header = {
        'referer': url,
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3',
    }
    
    # 회차 페이지 HTML 요청 및 파싱
    res = requests.get(url, headers=req_header)
    if res.ok:
        soup = BeautifulSoup(res.text, 'html.parser')
        # img_tag 추출 (wt_viewer 클래스 내부의 img 태그)
        img_tags = soup.select('.wt_viewer img')

    # image URL 추출
    img_url_list = [] 
    for img_tag in img_tags:
        img_url = img_tag['src']
        img_url_list.append(img_url)
            
        # 이미지 데이터 요청
        img_res = requests.get(img_url, headers=req_header)
        img_data = img_res.content
        file_path = os.path.join(save_dir, os.path.basename(img_url))
            
            # 바이너리 파일 저장
        with open(file_path, 'wb') as file:
            file.write(img_data)
            print(f"다운로드 완료: {file_path}")


# --- 호출 ---
download_one_episode('일렉시드', 341, 'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed')
```
