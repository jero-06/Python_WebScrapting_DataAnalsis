## 1-1. Daum 뉴스기사 제목 스크래핑하기 
* 아래의 url에서 뉴스기사의 링크와 제목을 출력하기
    - 다음 경제 뉴스 URL
    url = 'https://news.daum.net/economy'

```python
# requests, BeautifulSoup 모듈 import
import requests
import bs4
from bs4 import BeautifulSoup

# 경제 뉴스 섹션 URL
req_param = {
    'sid': 'economy'
}
url = 'https://news.daum.net/{sid}'.format(**req_param)
print(url)

# 요청 헤더 설정
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
}

# requests 모듈을 이용하여 HTTP GET 요청
res = requests.get(url, headers=req_header)
print(type(res))
print(res.status_code)

if res.ok:
    # response에서 text 추출
    html = res.text
    # BeautifulSoup 객체 생성
    soup = BeautifulSoup(html, 'html.parser')
    # print(soup)
# CSS 선택자
# print(soup.select("ul.list_newsheadline2 a[href*='mnews/article']"))
a_tags = soup.select("ul.list_newsheadline2 a[href*='https://v.daum.net/v/']")
print(type(a_tags))

# <a> 태그에서 href 속성 추출
for li_tag in soup.select('ul.list_newsheadline2 li'):
    a_tag = li_tag.select_one('a')
    link = a_tag['href']            #<a href="link">tile</a>
    # <strong> 태그에서 뉴스 제목 추출
    strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
    title = strong_tag.text         #<strong>title</strong>
    print(link, title)
```

    > 사회 section 선택



```python
import requests
from bs4 import BeautifulSoup

section_dict = {'기후/환경':'climate','사회':'society','경제':'economy','정치':'politics',\
                '국제':'world','문화':'culture','생활':'life','IT/과학':'tech','인물':'people'}

# 함수선언
def print_news(section_name):   # print_news('사회')
    sid = section_dict.get(section_name, 'society')
    url = f'https://news.daum.net/{sid}'
    print(f'======> {url} {section_name} 뉴스 <======')

    # 요청 헤더 설정
    req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }

    # requests 모듈을 이용하여 HTTP GET 요청
    res = requests.get(url, headers=req_header)

    if res.ok:
        # response에서 text 추출
        html = res.text
        # BeautifulSoup 객체 생성
        soup = BeautifulSoup(html, 'html.parser')
        # print(soup)
        # CSS 선택자
        # print(soup.select("ul.list_newsheadline2 a[href*='mnews/article']"))
        a_tags = soup.select("ul.list_newsheadline2 a[href*='https://v.daum.net/v/']")

        # <a> 태그에서 href 속성 추출
        for li_tag in soup.select('ul.list_newsheadline2 li'):
            a_tag = li_tag.select_one('a')
            link = a_tag['href']            #<a href="link">tile</a>
            # <strong> 태그에서 뉴스 제목 추출
            strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
            title = strong_tag.text         #<strong>title</strong>
            print(link, title)
# 함수호출
print_news('사회')
```
