## 1-1. Daum 뉴스기사 제목 스크래핑하기 

```python
# requests, BeautifulSoup 모듈 import
import requests
import bs4
from bs4 import BeautifulSoup

# 경제 뉴스 섹션 URL
req_param = {
    'sid': 'economy'
}
url = 'https://news.daum.net/{sid}'.format(**req_param)
print(url)

# 요청 헤더 설정
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
}

# requests 모듈을 이용하여 HTTP GET 요청
res = requests.get(url, headers=req_header)
print(type(res))
print(res.status_code)

if res.ok:
    # response에서 text 추출
    html = res.text
    # BeautifulSoup 객체 생성
    soup = BeautifulSoup(html, 'html.parser')
    # print(soup)
# CSS 선택자
# print(soup.select("ul.list_newsheadline2 a[href*='mnews/article']"))
a_tags = soup.select("ul.list_newsheadline2 a[href*='https://v.daum.net/v/']")
print(type(a_tags))

# <a> 태그에서 href 속성 추출
for li_tag in soup.select('ul.list_newsheadline2 li'):
    a_tag = li_tag.select_one('a')
    link = a_tag['href']

    strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
    title = strong_tag.text
    print(link, title)
```
