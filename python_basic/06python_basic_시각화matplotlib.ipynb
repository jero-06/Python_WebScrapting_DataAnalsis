## 0. 한글 폰트 설정 (필수)

설정하지 않으면 그래프의 한글이 네모(□)로 깨집니다.
Windows 는 '맑은 고딕(Malgun Gothic)' 을 사용합니다.

```python
import matplotlib.pyplot as plt

# 한글 폰트 + 마이너스 기호 깨짐 방지 (분석 노트북 맨 위에 항상 넣는 코드)
plt.rcParams['font.family'] = 'Malgun Gothic'   # macOS: 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
print('폰트 설정 완료')
```

## 1. 기본 구조 — Figure 와 Axes

- **Figure(fig)**: 그림 전체를 담는 도화지
- **Axes(ax)**: 실제 그래프가 그려지는 영역

`fig, ax = plt.subplots()` 로 만든 뒤 `ax` 에 그리는 방식이 표준입니다.

```python
# figsize=(가로, 세로) 인치 단위로 크기 지정
fig, ax = plt.subplots(figsize=(6, 4))

ax.plot([1, 2, 3, 4], [10, 20, 15, 30])   # x, y 데이터
ax.set_title('가장 기본적인 그래프')        # 제목
ax.set_xlabel('x축')
ax.set_ylabel('y축')
plt.show()   # 그래프 표시
```

## 2. 선 그래프 (plot) — 시간에 따른 변화

연도별 추세처럼 '흐름' 을 볼 때 사용합니다. (K-Pop 연도별 데뷔 추세가 이 형태)

```python
years = [2018, 2019, 2020, 2021, 2022]
male = [35, 83, 119, 46, 87]
female = [81, 73, 81, 74, 92]

fig, ax = plt.subplots(figsize=(9, 5))
# marker='o' : 각 점에 동그라미 표시 / label : 범례에 표시될 이름
ax.plot(years, male, marker='x', color='#2196F3', label='남성')
ax.plot(years, female, marker='o', color='#FF4081', label='여성')

ax.set_title('연도별 데뷔 인원 (예시)')
ax.set_xlabel('연도')
ax.set_ylabel('인원 수')
ax.legend()   # 범례 표시 (label 이 있어야 나옴)
plt.show()
```

## 3. 막대 그래프 (bar) — 항목별 비교

'소속사별 인원' 처럼 범주를 비교할 때 사용합니다.
값이 큰 항목이 많으면 **가로 막대(barh)** 가 이름 읽기에 좋습니다.

```python
companies = ['SM', 'JYP', 'FNC', 'Starship', 'YG']
counts = [58, 56, 52, 44, 41]

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# 세로 막대
axes[0].bar(companies, counts, color='steelblue')
axes[0].set_title('세로 막대 (bar)')

# 가로 막대 (barh) — 이름이 길 때 유리
axes[1].barh(companies, counts, color='seagreen')
axes[1].set_title('가로 막대 (barh)')
axes[1].invert_yaxis()   # 큰 값이 위로 오도록 순서 뒤집기

plt.tight_layout()   # 그래프 간 간격 자동 조정
plt.show()
```

## 4. 산점도 (scatter) — 두 값의 관계

'키와 몸무게' 처럼 두 수치의 상관관계를 볼 때 사용합니다.

```python
heights = [164, 178, 170, 160, 182, 168, 175]
weights = [48, 62, 55, 45, 68, 52, 60]

fig, ax = plt.subplots(figsize=(7, 5))
# alpha : 투명도 (겹친 점 파악에 유용)
ax.scatter(heights, weights, color='tomato', alpha=0.7, s=80)
ax.set_title('키 vs 몸무게')
ax.set_xlabel('키 (cm)')
ax.set_ylabel('몸무게 (kg)')
plt.show()
```

## 5. 히스토그램 (hist) — 분포 확인

'데뷔 나이 분포' 처럼 값이 어디에 몰려 있는지 볼 때 사용합니다.
`bins` 는 막대(구간)의 개수입니다.

```python
import numpy as np

# 평균 19.6, 표준편차 2.9 인 가상의 데뷔 나이 500개 생성
np.random.seed(0)   # 매번 같은 난수가 나오도록 고정
debut_ages = np.random.normal(19.6, 2.9, 500)

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(debut_ages, bins=20, color='mediumpurple', edgecolor='white')

# 평균선 추가 (axvline: 세로 기준선)
ax.axvline(debut_ages.mean(), color='red', linestyle='--', label=f'평균 {debut_ages.mean():.1f}세')
ax.set_title('데뷔 나이 분포 (예시)')
ax.set_xlabel('나이')
ax.set_ylabel('빈도')
ax.legend()
plt.show()
```

## 6. 파이 차트 (pie) — 비율

'콘텐츠 유형 비율' 처럼 전체 대비 비중을 볼 때 사용합니다.

```python
labels = ['Movie', 'TV Show']
sizes = [6131, 2676]

fig, ax = plt.subplots(figsize=(6, 6))
# autopct : 각 조각에 퍼센트 표시 / startangle : 시작 각도
ax.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90,
       colors=['#E50914', '#221F1F'], textprops={'color': 'white'})
ax.set_title('콘텐츠 유형 비율 (예시)')
plt.show()
```

## 7. 여러 그래프 나란히 — subplots

`plt.subplots(행, 열)` 로 여러 그래프를 한 화면에 배치합니다.
반환된 `axes` 는 그래프들의 배열입니다.

```python
# 2행 2열 = 그래프 4개
fig, axes = plt.subplots(2, 2, figsize=(11, 8))

axes[0, 0].plot([1, 2, 3], [1, 4, 9]);      axes[0, 0].set_title('선')
axes[0, 1].bar(['A', 'B', 'C'], [3, 7, 5]); axes[0, 1].set_title('막대')
axes[1, 0].scatter([1, 2, 3], [3, 1, 2]);   axes[1, 0].set_title('산점도')
axes[1, 1].hist(np.random.randn(200));      axes[1, 1].set_title('히스토그램')

fig.suptitle('한 화면에 4개 그래프', fontsize=14)   # 전체 제목
plt.tight_layout()
plt.show()
```

## 8. seaborn 으로 더 쉽게

`seaborn` 은 matplotlib 을 기반으로 더 예쁘고 간단하게 그려주는 라이브러리입니다.
특히 **DataFrame 을 바로 넣을 수 있어** 분석에서 자주 씁니다.

```python
import seaborn as sns
import pandas as pd

sns.set_theme(style='whitegrid', font='Malgun Gothic')

# DataFrame 준비
df = pd.DataFrame({
    'company': companies,
    'count': counts,
})

fig, ax = plt.subplots(figsize=(9, 4))
# x, y 에 컬럼명만 주고 data=df 로 DataFrame 전달 (매우 편리)
sns.barplot(data=df, x='count', y='company', ax=ax, hue='company')#palette='Blues_r')
ax.set_title('seaborn 으로 그린 소속사별 인원')
plt.show()
```

```python
# 그래프를 이미지 파일로 저장하기 (보고서/발표용)
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(years, female, marker='o', color='#FF4081')
ax.set_title('저장 예시')

# dpi : 해상도 / bbox_inches='tight' : 여백 최소화
fig.savefig('sample_chart.png', dpi=100, bbox_inches='tight')
print('sample_chart.png 저장 완료')
plt.close(fig)   # 저장 후 메모리 정리
```

## 정리
- **항상 먼저**: 한글 폰트 설정 (`Malgun Gothic`)
- **fig, ax = plt.subplots()** 로 시작 → `ax` 에 그리기
- 그래프 종류: `plot`(추세) · `bar`(비교) · `scatter`(관계) · `hist`(분포) · `pie`(비율)
- **제목/축/범례**: `set_title`, `set_xlabel`, `legend`
- **subplots(행,열)** 로 여러 개 배치, `tight_layout()` 로 정리
- **seaborn** 은 DataFrame 을 바로 그려 편리
- **savefig** 로 이미지 저장
